### Question 4 (Easy) — `total_cost`

### Question 5 (Easy) — `item_total` and `cart_total`

Write two functions:

- `cart_total(15000, 2, 8000, 3)` ➞ `54000`

### Question 6 (Easy) — `simple_discount_amount`

Write a function `simple_discount_amount(price, discount_percent=10)` that returns the discount amount (not the discounted price): `price * discount_percent / 100`. Default the discount to 10%.

- `simple_discount_amount(50000)` ➞ `5000.0`

### Question 7 (Easy) — `circle_area_approx`

Write a function `circle_area_approx(radius)` that returns the approximate area of a circle, using `3.14` for pi: `3.14 * radius ** 2`.

- `circle_area_approx(2)` ➞ `12.56`

### Question 8 (Medium) — `grade_percentage`

Write a function `grade_percentage(points_earned, points_possible)` that returns the score as a percentage: `points_earned / points_possible * 100`.

- `grade_percentage(45, 50)` ➞ `90.0`
- `grade_percentage(63, 90)` ➞ `70.0`

### Question 9 (Medium) — `distance_with_acceleration`

Write a function `distance_with_acceleration(initial_speed, time, acceleration)` that returns distance traveled under constant acceleration: `initial_speed * time + 0.5 * acceleration * time ** 2`.

- `distance_with_acceleration(5, 2, 3)` ➞ `16.0`
- `distance_with_acceleration(10, 3, 4)` ➞ `48.0`

```python
# YOUR CODE HERE
def distance_with_acceleration(initial_speed, time, acceleration):
    return initial_speed * time + 0.5 * acceleration * time ** 2

print(distance_with_acceleration(5, 2, 3))
print(distance_with_acceleration(10, 3, 4))
```

### Question 10 — Challenge (Hard) — `compound_interest`

Write a function `compound_interest(principal, rate_percent, years)` that returns the **interest earned** (not the total amount) after compounding once per year: `principal * (1 + rate_percent / 100) ** years - principal`.

- `compound_interest(100000, 10, 2)` ➞ about `21000.0`
- `compound_interest(500000, 5, 3)` ➞ about `78812.5`

```python
# YOUR CODE HERE
def compound_interest(principal, rate_percent, years):
  return principal * (1 + rate_percent / 100) ** years - principal

print(compound_interest(100000, 10, 2))
print(compound_interest(500000, 5, 3))
```
